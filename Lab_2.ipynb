```python
def func8(A):
    # func8 initiates a recursive selection sort on the input array A.
    # Time Complexity: O(N^2), where N is the length of the array.
    func8_helper(A, 0, len(A) - 1)
    return A

def func8_helper(A, i, j):
    # func8_helper recursively sorts a subarray A[i...j] using selection sort logic.
    # It finds the minimum element in the unsorted portion and places it at the current position.
    # Time Complexity: O(N^2) in the worst case for the subarray it processes, contributing to O(N^2) overall.
    if i >= j:
        return

    # Find the index of the minimum element in the unsorted part A[i...j]
    min_idx = i
    for k in range(i + 1, j + 1):
        if A[k] < A[min_idx]:
            min_idx = k

    # Swap the found minimum element with the element at the current position i
    A[i], A[min_idx] = A[min_idx], A[i]

    # Recursively sort the remainder of the array, starting from i+1
    func8_helper(A, i + 1, j)
```

```python
def func8(A):
    # func8 implements the Merge Sort algorithm to sort an array A.
    # It recursively divides the array into halves, sorts them, and then merges the sorted halves.
    # Time Complexity: O(N log N), where N is the length of the array.

    # Base case: if the array has 0 or 1 element, it's already sorted
    if len(A) <= 1:
        return A

    # Divide step: find the middle and split the array into two halves
    mid = len(A) // 2
    left_half = A[:mid]
    right_half = A[mid:]

    # Conquer step: recursively sort both halves
    sorted_left = func8(left_half)
    sorted_right = func8(right_half)

    # Combine step: merge the sorted halves
    return merge_arrays(sorted_left, sorted_right)

def merge_arrays(left, right):
    # merge_arrays takes two sorted arrays (left and right) and merges them into a single sorted array.
    # Time Complexity: O(N), where N is the total number of elements in both left and right arrays.
    merged_array = []
    i = 0  # pointer for left array
    j = 0  # pointer for right array

    # Compare elements from both arrays and add the smaller one to merged_array
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            merged_array.append(left[i])
            i += 1
        else:
            merged_array.append(right[j])
            j += 1

    # Append any remaining elements from the left array
    while i < len(left):
        merged_array.append(left[i])
        i += 1

    # Append any remaining elements from the right array
    while j < len(right):
        merged_array.append(right[j])
        j += 1

    return merged_array
```
